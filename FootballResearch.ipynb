```python
import cloudscraper
import pandas as pd
from bs4 import BeautifulSoup

# Standard career performance URLs (without /plus/1 breakdown)
players_urls = {
    "R9": "https://www.transfermarkt.us/ronaldo/leistungsdaten/spieler/3140",
    "Henry": "https://www.transfermarkt.us/thierry-henry/leistungsdaten/spieler/3207",
    "Shevchenko": "https://www.transfermarkt.us/andriy-shevchenko/leistungsdaten/spieler/3522",
    "Crespo": "https://www.transfermarkt.us/hernan-crespo/leistungsdaten/spieler/3410",
    "Shearer": "https://www.transfermarkt.us/alan-shearer/leistungsdaten/spieler/3110",
    "Cole": "https://www.transfermarkt.us/andy-cole/leistungsdaten/spieler/3238",
    "Vieri": "https://www.transfermarkt.us/christian-vieri/leistungsdaten/spieler/3517",
    "Owen": "https://www.transfermarkt.us/michael-owen/leistungsdaten/spieler/3134",
    "Del Piero": "https://www.transfermarkt.us/alessandro-del-piero/leistungsdaten/spieler/4289",
    "Raul": "https://www.transfermarkt.us/raul/leistungsdaten/spieler/7349",
    "Floyd Hasselbank": "https://www.transfermarkt.us/jimmy-floyd-hasselbaink/leistungsdaten/spieler/3158",
    "Batigol": "https://www.transfermarkt.us/gabriel-batistuta/leistungsdaten/spieler/5959",
    "Fowler": "https://www.transfermarkt.us/robbie-fowler/leistungsdaten/spieler/3125",
    "Ferdinand": "https://www.transfermarkt.us/les-ferdinand/leistungsdaten/spieler/3281",
    "Klinsmann": "https://www.transfermarkt.us/jurgen-klinsmann/leistungsdaten/spieler/16980",
    "Suker": "https://www.transfermarkt.us/davor-suker/leistungsdaten/spieler/8013",
    "Morientes": "https://www.transfermarkt.us/fernando-morientes/leistungsdaten/spieler/7531",
    "Makaay": "https://www.transfermarkt.us/roy-makaay/leistungsdaten/spieler/3872",
    "Kluivert": "https://www.transfermarkt.us/patrick-kluivert/leistungsdaten/spieler/3702",
    "Bierhoff": "https://www.transfermarkt.us/oliver-bierhoff/leistungsdaten/spieler/85",
}

scraper = cloudscraper.create_scraper()


def clean_num(val_str):
    """Strips formatting characters and returns integer."""
    if not val_str:
        return 0
    clean = (
        str(val_str)
        .replace(".", "")
        .replace("'", "")
        .replace("’", "")
        .replace("-", "0")
        .strip()
    )
    return int(clean) if clean.isdigit() else 0


def get_best_5_seasons(url):
    response = scraper.get(url)
    if response.status_code != 200:
        return [0, 0, 0, 0, 0]

    soup = BeautifulSoup(response.text, "html.parser")
    table = soup.find("table", class_="items")
    if not table:
        return [0, 0, 0, 0, 0]

    tbody = table.find("tbody")
    if not tbody:
        return [0, 0, 0, 0, 0]

    mins_per_gi_list = []

    for tr in tbody.find_all("tr", recursive=False):
        cells = tr.find_all("td")
        if len(cells) < 6:
            continue

        texts = [c.get_text(strip=True) for c in cells]

        # On standard career pages:
        # Goals = 3rd from end, Assists = 2nd from end, Minutes = last cell
        goals = clean_num(texts[-3])
        assists = clean_num(texts[-2])
        mins = clean_num(texts[-1])

        gi = goals + assists
        if gi > 0 and mins > 500:  # Ignore low-minute anomaly seasons
            mins_per_gi_list.append(round(mins / gi))

    if len(mins_per_gi_list) < 5:
        mins_per_gi_list.extend([0] * (5 - len(mins_per_gi_list)))
        return mins_per_gi_list[:5]

    # Find 5-season contiguous stretch with lowest average Mins per G/A
    best_avg = float("inf")
    best_window = []

    for i in range(len(mins_per_gi_list) - 4):
        window = mins_per_gi_list[i : i + 5]
        w_avg = sum(window) / 5
        if w_avg < best_avg:
            best_avg = w_avg
            best_window = window

    return best_window


scraped_data = {}
for player, url in players_urls.items():
    print(f"Scraping {player}...")
    scraped_data[player] = get_best_5_seasons(url)

df = pd.DataFrame.from_dict(
    scraped_data, orient="index", columns=["S1", "S2", "S3", "S4", "S5"]
)
print("\n--- Corrected Scraped Results ---")
print(df)
```
