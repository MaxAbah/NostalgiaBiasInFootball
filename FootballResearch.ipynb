```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Dataset
data_by_striker = {
    'R9':               [71, 107, 109, 130, 84],
    'Henry':            [92, 123, 101, 83, 78],
    'Shevchenko':       [96, 114, 141, 233, 112],
    'Crespo':           [104, 110, 92, 122, 92],
    'Shearer':          [87, 102, 85, 93, 98],
    'Cole':             [74, 117, 166, 103, 110],
    'Vieri':            [116, 89, 128, 80, 101],
    'Owen':             [99, 111, 139, 97, 98],
    'Del Piero':        [85, 95, 110, 105, 92],
    'Raul':             [90, 105, 115, 98, 88],
    'Floyd Hasselbank': [138, 127, 105, 97, 109],
    'Batigol':          [107, 127, 193, 104, 120],
    'Fowler':           [121, 104, 105, 143, 132],
    'Ferdinand':        [128, 132, 101, 111, 109],
    'Klinsmann':        [168, 152, 103, 118, 120],
    'Suker':            [128, 146, 164, 101, 123],
    'Morientes':        [129, 99, 130, 119, 112],
    'Makaay':           [112, 106, 156, 97, 111],
    'Kluivert':         [104, 98, 117, 106, 133],
    'Bierhoff':         [164, 140, 91, 101, 152]
}

fig, ax = plt.subplots(figsize=(14, 8), facecolor='#FAFAFA')
ax.set_facecolor('#FAFAFA')

periods = np.array([1, 2, 3, 4, 5])

# 3. Summary statistics on RAW values (25th–75th percentile band + median)
all_raw_data = np.array(list(data_by_striker.values()))
median_curve = np.median(all_raw_data, axis=0)
p25_curve = np.percentile(all_raw_data, 25, axis=0)
p75_curve = np.percentile(all_raw_data, 75, axis=0)

# Shaded background band
ax.fill_between(periods, p25_curve, p75_curve, color='#E5E7EB', alpha=0.8, label='25th–75th Percentile Range')
ax.plot(periods, median_curve, color='#9CA3AF', linestyle='--', linewidth=1.5, label='Cohort Median')

# 4. Define Highlighted Players & Colors
highlight_players = {
    'Henry': '#1E40AF',    
    'Raul': '#EA580C',  
    'Shevchenko': '#16A34A', 
    'R9': '#DC2626',         
    'Owen': '#8B5CF6'     
}

# Plot non-highlighted players in subtle gray lines
for player, values in data_by_striker.items():
    if player not in highlight_players:
        ax.plot(periods, values, color='#D1D5DB', linewidth=1.2, alpha=0.6)

# Plot highlighted players with vibrant lines & markers
for player, color in highlight_players.items():
    values = data_by_striker[player]
    ax.plot(periods, values, color=color, linewidth=2.5, marker='o', markersize=4, label=player)
    
    # Direct line label on the right side of the plot
    ax.text(
        periods[-1] + 0.05, values[-1], f' {player}',
        color=color, fontweight='bold', fontsize=10, va='center'
    )

ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)
ax.spines['bottom'].set_color('#9CA3AF')

ax.yaxis.grid(True, linestyle='-', color='#E5E7EB', linewidth=1)
ax.set_axisbelow(True)

# Titles & Axis Labels
plt.title('90s–00s Strikers — Performance Progression Across Groups', fontsize=16, fontweight='bold', pad=20, loc='left', color='#111827')
ax.set_xlabel('Group / Period', fontsize=11, fontweight='bold', color='#374151', labelpad=10)
ax.set_ylabel('G/A per 90', fontsize=11, fontweight='bold', color='#374151', labelpad=10)

ax.set_xticks(periods)
ax.set_xlim(0.9, 5.6)  

ax.legend(loc='lower center', bbox_to_anchor=(0.5, -0.18), ncol=4, frameon=False, fontsize=9)

plt.tight_layout()
plt.show()
```
